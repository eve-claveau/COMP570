```python
import csv
import matplotlib.pyplot as plt
import os 

os.chdir('/home/eve/comp570github/COMP570')
with open("results.csv", "r") as fi:
    reader = csv.reader(fi)
    dicti = {}
    header = next(reader)
    for line in reader:
        if line[0] == "HEAT/HOT WATER":
            dicti[line[1]] = int(line[2])

plt.bar(dicti.keys(), dicti.values())

plt.title("Heat Complaint Count per Borough Jan/Feb 2024")
plt.xlabel("Borough")
plt.ylabel("Count")

plt.show
```
